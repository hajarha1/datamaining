```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

# نعمل نسخة لتشفيرها حتى لا نغيّر df الأصلية
df_encoded = df.copy()

# الأعمدة المستخدمة في النموذج
features = ['ProductID', 'ProductName', 'Category', 'Price', 'Brand', 'Rating',
            'Stock', 'WarrantyYears', 'SupplierCountry', 'WeightKg', 'VolumeCm3',
            'PowerWatt', 'MaterialType', 'UsageType', 'PriceCategory']

# نحتاج لتشفير كل الأعمدة النصية (بما فيها y)
label_encoders = {}
for col in features + ['ConnectivityType']:
    if df_encoded[col].dtype == 'object' or df_encoded[col].isnull().any():
        le = LabelEncoder()
        df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))
        label_encoders[col] = le

# تقسيم البيانات
train_data = df_encoded[df['ConnectivityType'].notnull()]
test_data = df_encoded[df['ConnectivityType'].isnull()]

X_train = train_data[features]
y_train = train_data['ConnectivityType']
X_test = test_data[features]

# تدريب النموذج
model = RandomForestClassifier()
model.fit(X_train, y_train)

# التنبؤ بالقيم المفقودة
predicted_encoded = model.predict(X_test)

# فك التشفير لعكس القيم الرقمية إلى النصوص الأصلية
decoded_preds = label_encoders['ConnectivityType'].inverse_transform(predicted_encoded)

# تعبئة القيم الأصلية
df.loc[df['ConnectivityType'].isnull(), 'ConnectivityType'] = decoded_preds
```

```python

numeric_df = df.select_dtypes(include='number')

# حساب الربع الأول والثالث
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

# حساب IQR
IQR = Q3 - Q1

# تحديد الحدود الدنيا والعليا
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = (numeric_df < lower) | (numeric_df > upper)

print(outliers.sum())
```

```python
# استخراج الصفوف التي تحتوي على قيم شاذة في Price أو PowerWatt
rows_with_outliers = outliers[['Price', 'PowerWatt']].any(axis=1)
outlier_rows = df[rows_with_outliers]

print(outlier_rows)
```

```python
from sklearn.cluster import KMeans
cols = df.select_dtypes(include='number').columns  # استخراج أسماء الأعمدة الرقمية
X = df[cols].values                                # استخراج القيم الرقمية نفسها

ilist=[]
n=10
for i in range(1,n):
    Kmeans=KMeans(n_clusters=i,random_state=42,n_init='auto')
    Kmeans.fit(X)
    ilist.append(Kmeans.inertia_)
```

```python
plt.plot(range(1,10),ilist)
plt.title('Elbow')
```

```python
Kmeans=KMeans(n_clusters=3,random_state=42)
y_kmeans=Kmeans.fit_predict(X)
df['cluster']=y_kmeans
df
```

```python
import matplotlib.pyplot as plt

plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s=10, c='r', label="cluster 1")
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s=10, c='b', label="cluster 2")
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s=10, c='g', label="cluster 3")
# إذا كان لديك cluster 4 (الفهرس 3) يمكن تفعيله:
# plt.scatter(X[y_kmeans == 3, 0], X[y_kmeans == 3, 1], s=10, c='m', label="cluster 4")

# رسم مراكز المجموعات
plt.scatter(Kmeans.cluster_centers_[:, 0], Kmeans.cluster_centers_[:, 1], s=100, c='y', label='Centroids')

plt.title('Clusters of customers')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```

```python
import pandas as pd

invoices_df = pd.read_csv('Invoices_Dataset_for_Association_Rules.csv')

# عرض أول 5 صفوف
invoices_df.head()
```

```python
invoices_df.isnull().sum()
```

```python
invoices_df.describe()
```

```python
orders=invoices_df["InvoiceID"].unique()
orders

dataset = []
for order_id in orders:
    lstinner = []
    for j in range(invoices_df.shape[0]):
        if order_id == invoices_df.iloc[j]["InvoiceID"]:
            lstinner.append(invoices_df.iloc[j]["ProductID"])
    dataset.append(lstinner)

# عرض النتيجة
dataset
```

```python
from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_array = te.fit(dataset).transform(dataset)

# تحويلها إلى DataFrame
df_encoded = pd.DataFrame(te_array, columns=te.columns_)

df_encoded.head()
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
frequent_itemset = apriori(df_encoded, min_support=0.03, use_colnames=True)
print(frequent_itemset)
```

```python
rules = association_rules(frequent_itemset, metric="confidence", min_threshold=0.70)
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
sorted_rules=rules.sort_values(by=['confidence','lift'],ascending=[False,False])
```

```python
rules_list_single = []

for idx, rule in sorted_rules.iterrows():
    antecedents = list(rule['antecedents'])   # frozenset → list مباشرة
    consequents = list(rule['consequents'])   # frozenset → list مباشرة
    support = rule['support']
    confidence = rule['confidence']
    lift = rule['lift']

    for ant in antecedents:
        for cons in consequents:
            rules_list_single.append({
                'AntecedentProductID': ant,
                'ConsequentProductID': cons,
                'Support': support,
                'Confidence': confidence,
                'Lift': lift
            })

rules_df_single = pd.DataFrame(rules_list_single)
```

```python
# تحويل القائمة إلى DataFrame
rules_df = pd.DataFrame(rules_df_single)

# عرض DataFrame
print(rules_df_single)
```
