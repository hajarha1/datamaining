```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

# نعمل نسخة لتشفيرها حتى لا نغيّر df الأصلية
df_encoded = df.copy()

# الأعمدة المستخدمة في النموذج
features = ['ProductID', 'ProductName', 'Category', 'Price', 'Brand', 'Rating',
            'Stock', 'WarrantyYears', 'SupplierCountry', 'WeightKg', 'VolumeCm3',
            'PowerWatt', 'MaterialType', 'UsageType', 'PriceCategory']

# نحتاج لتشفير كل الأعمدة النصية (بما فيها y)
label_encoders = {}
for col in features + ['ConnectivityType']:
    if df_encoded[col].dtype == 'object' or df_encoded[col].isnull().any():
        le = LabelEncoder()
        df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))
        label_encoders[col] = le

# تقسيم البيانات
train_data = df_encoded[df['ConnectivityType'].notnull()]
test_data = df_encoded[df['ConnectivityType'].isnull()]

X_train = train_data[features]
y_train = train_data['ConnectivityType']
X_test = test_data[features]

# تدريب النموذج
model = RandomForestClassifier()
model.fit(X_train, y_train)

# التنبؤ بالقيم المفقودة
predicted_encoded = model.predict(X_test)

# فك التشفير لعكس القيم الرقمية إلى النصوص الأصلية
decoded_preds = label_encoders['ConnectivityType'].inverse_transform(predicted_encoded)
df.loc[df['ConnectivityType'].isnull(), 'ConnectivityType'] = decoded_preds
```

```python
numeric_df = df.select_dtypes(include='number')

# حساب الربع الأول والثالث
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

# حساب IQR
IQR = Q3 - Q1

# تحديد الحدود الدنيا والعليا
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = (numeric_df < lower) | (numeric_df > upper)

print(outliers.sum())
```

```python

for col in numeric_df.columns:
    median = numeric_df[col].median()
    # استبدال القيم الأقل من الحد الأدنى أو الأكبر من الحد الأعلى بالوسيط
    df.loc[(df[col] < lower[col]) | (df[col] > upper[col]), col] = median
```
